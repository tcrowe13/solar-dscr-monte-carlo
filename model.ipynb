```python
import pandas as pd
import numpy as np

# Assumptions
years = list(range(1, 21))
p50_mwh = 100_000
tariff = 60            # £ per MWh
opex = 1_500_000       # £ per year
debt = 25_000_000      # £ initial debt
rate = 0.06            # 6% interest
tenor = 18             # years of debt repayment

# Operating cash flow
df = pd.DataFrame({"year": years})
df["revenue"] = p50_mwh * tariff
df["opex"] = opex
df["cfads"] = df["revenue"] - df["opex"]

# Annuity debt service: a level payment each year over the tenor
annuity = debt * rate / (1 - (1 + rate) ** -tenor)
df["debt_service"] = np.where(df["year"] <= tenor, annuity, 0)

# DSCR
df["dscr"] = np.where(df["debt_service"] > 0, df["cfads"] / df["debt_service"], np.nan)

df.round(2)
```

```python
# Solar degradation: output falls a bit each year
degradation = 0.005    # 0.5% per year

df["p50_mwh"] = p50_mwh * (1 - degradation) ** (df["year"] - 1)

# P90 from P50: assume annual generation is normally distributed
# P90 = P50 * (1 - 1.2816 * sigma), where 1.2816 is the z-score for 90%
sigma = 0.08           # 8% standard deviation (typical-ish for one-year solar yield)
z90 = 1.2816
df["p90_mwh"] = df["p50_mwh"] * (1 - z90 * sigma)

# Rebuild cash flows and DSCR for both cases
for case in ["p50", "p90"]:
    df[f"cfads_{case}"] = df[f"{case}_mwh"] * tariff - opex
    df[f"dscr_{case}"] = np.where(
        df["debt_service"] > 0, df[f"cfads_{case}"] / df["debt_service"], np.nan
    )

print(df[["year", "cfads_p50", "cfads_p90", "dscr_p50", "dscr_p90"]].round(2).to_string())
print("Min DSCR P50:", df["dscr_p50"].min().round(2))
print("Min DSCR P90:", df["dscr_p90"].min().round(2))
```

```python
import matplotlib.pyplot as plt

np.random.seed(42)          # makes results repeatable
n_sims = 10_000

# Annual generation factor: 1.0 = P50. Each year is an independent draw here.
# (Simplification: in reality, yield uncertainty is partly correlated across years.)
factors = np.random.normal(loc=1.0, scale=sigma, size=(n_sims, len(df)))

# Generation for every simulation and year
gen = df["p50_mwh"].values * factors            # shape (10000, 20)
cfads_sim = gen * tariff - opex

# DSCR only in debt years
ds = df["debt_service"].values
mask = ds > 0
dscr_sim = cfads_sim[:, mask] / ds[mask]

min_dscr = dscr_sim.min(axis=1)                 # worst year in each simulation

# Results
lockup = 1.20
default = 1.05
print("Mean min DSCR:", min_dscr.mean().round(2))
print("5th percentile min DSCR:", np.percentile(min_dscr, 5).round(2))
print(f"P(min DSCR < {lockup}):", (min_dscr < lockup).mean().round(4))
print(f"P(min DSCR < {default}):", (min_dscr < default).mean().round(4))

# Plot
plt.figure(figsize=(9, 5))
plt.hist(min_dscr, bins=60)
plt.axvline(lockup, linestyle="--", label=f"Lock-up {lockup}x")
plt.axvline(default, linestyle="--", label=f"Default {default}x")
plt.axvline(df["dscr_p90"].min(), linestyle=":", label="Deterministic P90 min")
plt.xlabel("Minimum DSCR across debt life")
plt.ylabel("Number of simulations")
plt.title("Monte Carlo: minimum DSCR distribution")
plt.legend()
plt.show()
```

```python
def run_sim(corr_share, n_sims=10_000, seed=42):
    """corr_share = share of variance that is common to all years (0 to 1)"""
    rng = np.random.default_rng(seed)
    sigma_c = sigma * np.sqrt(corr_share)
    sigma_i = sigma * np.sqrt(1 - corr_share)
    common = rng.normal(0, sigma_c, size=(n_sims, 1))
    indep = rng.normal(0, sigma_i, size=(n_sims, len(df)))
    gen = df["p50_mwh"].values * (1 + common + indep)
    cfads = gen * tariff - opex
    dscr = cfads[:, mask] / ds[mask]
    return dscr.min(axis=1)

rows = []
for share in [0, 0.25, 0.5, 0.75, 1.0]:
    m = run_sim(share)
    rows.append({
        "corr_share": share,
        "mean_min": m.mean(),
        "p5_min": np.percentile(m, 5),
        "P(<1.20)": (m < 1.20).mean(),
        "P(<1.05)": (m < 1.05).mean(),
    })

pd.DataFrame(rows).round(4)
```
